```python
import pandas as pd

july_weather = {
    "Day": list(range(1, 32)),
    "Max_Temp_C": [
        28.1, 27.9, 27.5, 27.3, 27.0, 26.8, 27.2, 27.5, 27.8, 28.0,
        27.4, 27.1, 26.9, 27.2, 27.6, 27.8, 28.2, 27.9, 27.4, 27.1,
        26.8, 27.3, 27.7, 28.0, 28.1, 27.9, 27.5, 27.2, 27.6, 28.0, 28.3
    ],
    "Min_Temp_C": [
        24.0, 23.8, 23.5, 23.2, 23.1, 23.0, 23.2, 23.4, 23.6, 23.8,
        23.5, 23.3, 23.1, 23.2, 23.5, 23.7, 23.9, 23.8, 23.4, 23.2,
        23.0, 23.3, 23.5, 23.8, 23.9, 23.7, 23.5, 23.2, 23.4, 23.7, 23.9
    ],
    "Rainfall_mm": [
        35.2, 42.0, 50.1, 62.4, 48.0, 55.3, 40.2, 30.5, 25.1, 20.4,
        38.0, 45.6, 58.2, 41.0, 32.5, 28.0, 15.2, 22.1, 36.4, 49.0,
        52.1, 39.4, 27.8, 18.5, 24.0, 31.2, 40.5, 48.2, 33.0, 21.4, 18.0
    ]
}

df = pd.DataFrame(july_weather)


df["Avg_Temp_C"] = (df["Max_Temp_C"] + df["Min_Temp_C"]) / 2


max_temp = df["Max_Temp_C"].max()
min_temp = df["Min_Temp_C"].min()
overall_avg_temp = df["Avg_Temp_C"].mean()
total_rainfall = df["Rainfall_mm"].sum()


hottest_day = df.loc[df["Max_Temp_C"].idxmax()]["Day"]
coolest_day = df.loc[df["Min_Temp_C"].idxmin()]["Day"]


print("--- MANIPAL JULY WEATHER ANALYSIS ---")
print(f"Maximum Temperature: {max_temp:.1f} °C (Day {int(hottest_day)})")
print(f"Lowest Temperature:  {min_temp:.1f} °C (Day {int(coolest_day)})")
print(f"Average Temperature: {overall_avg_temp:.2f} °C")
print(f"Total Rainfall:      {total_rainfall:.1f} mm")
```

```python
# find the avg age of servivers and non-servivers

# passangers who are older than 60 yrs

# find the avg fair for each passanger

import pandas as pd


df = pd.read_csv("Titanic-Dataset.csv")

df.head()
```

```python
avg_age = df.groupby('Survived')['Age'].mean()

print("--- Average Age by Survival Status ---")
print(f"Non-Survivors (0): {avg_age[0]:.2f} years")
print(f"Survivors (1):     {avg_age[1]:.2f} years")
```

```python
passengers_over_60 = df[df["Age"] > 60]
print(passengers_over_60)
```
